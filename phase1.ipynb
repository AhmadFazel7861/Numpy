### Numpy

```python
import numpy as np
```

```python
arr_1d = np.array([1,2,3,4,5])

print(arr_1d)

arr_2d = np.array([[1,2,3,4],[6,7,8,9]])

print(arr_2d)
```

### List vs numpy arrays


```python
py_list = [1,2,3,4,5]
print(py_list * 2)

arr = np.array([1,2,3,4])
print(arr * 2)
```

### Creating arrays from scratch

```python
zeros = np.zeros((4,5))
print(zeros)
```

```python
ones = np.ones((2,3))
print(ones)
```

```python
full = np.full((1,2),7)
print(full)
```

```python
random = np.random.random((2,3))
print(random)
```

```python
sequence = np.arange(0,10,2)
print(sequence)
```

### Vector , Matrix , Tensor

```python
vector = np.array([1,2,3,4])
print(vector)
```

```python
matrix = np.array([[1,2,3],[4,5,6]])
print(matrix)
```

```python
tensor = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(tensor)
```

### Array's properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print('shape:', arr.shape)
print('Size:', arr.size)
print('Dimension:', arr.ndim)
print('Dtype:', arr.dtype)
```

array Reshaping

```python
original = np.arange(12)
print(original)
```

```python
reshaped= original.reshape((3,4))
print(reshaped)
```

```python
transpose = reshaped.T
print(transpose)
```

```python
flattened = reshaped.flatten()
print(flattened)
```

```python
raveled = reshaped.ravel()
print(raveled)
```
